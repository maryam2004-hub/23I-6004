# Task 9

```python
import pandas as pd
import sqlite3

# Read the raw CSV normally first
df_raw = pd.read_csv("adult.csv")

# Create/connect to a local SQLite database file
conn = sqlite3.connect("adult_income.db")

# Write the DataFrame into a SQL table called adult_income
df_raw.to_sql("adult_income", conn, if_exists="replace", index=False)

print("Table created. Original CSV shape:", df_raw.shape)

query = "SELECT * FROM adult_income WHERE age > 30;"
df_sql = pd.read_sql_query(query, conn)

print("Query result shape:", df_sql.shape)

# Cross-check: count rows manually with pandas filtering
expected = df_raw[df_raw["age"] > 30]
print("Pandas filter shape:", expected.shape)
print("SQL query shape:    ", df_sql.shape)

assert df_sql.shape == expected.shape, "Mismatch between SQL result and pandas filter!"
print("SQL extraction matches pandas filtering — extraction confirmed correct.")

conn.close()
```
